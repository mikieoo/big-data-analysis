```python
import networkx as nx

g1=nx.Graph()
g1.add_node('David')
g1.add_node('Julia')
g1.add_node('John')
g1.add_node('Mary')
g1.add_node('Juda')
g1.add_nodes_from(['Judy','Karen'])
g1.remove_node('Julia')

g1.add_edge('David','John')
g1.add_edge('David','Mary')
g1.add_edge('David','Karen')
g1.add_edges_from([('John','Mary'),('John','Judy')])
g1.remove_edge('Mary','John')

nx.draw(g1,with_labels=True,font_weight='bold')
```

```python
d=dict(g1.degree)
nx.draw(g1,nodelist=d.keys(),node_size=[v*1000 for v in d.values()],with_labels=True,font_weight='bold')
g1.nodes
```

```python
g2=nx.DiGraph()
g2.add_edges_from([(1,2),(1,3),(1,4),(3,4)])
nx.draw(g2,with_labels=True,font_weight='bold')
print(g2.in_degree)
print(g2.out_degree)
```

```python
class SoccerPlayer:
  def __init__(self, name, position, back_number):
    self.name = name
    self.position = position
    self.back_number = back_number

  def change_back_number(self, change_number):
    self.back_number = change_number

  def __str__(self):
    return f'이름: {self.name}, 포지션: {self.position}, 백넘버: {self.back_number}'

yangmin = SoccerPlayer("Lee","MF",10)
yangmin.change_back_number(11)
print(yangmin)
```

```python
import pandas as pd

data = {
    '메뉴 ID':['M1','M2','M3','M4','M5'],
    '메뉴명':['아메리카노','카페라떼','녹차라떼','딸기 스무디','초코라떼'],
    '카테고리':['커피','커피','차','스무디','차'],
    '가격':[3000,3500,4000,4500,3800],
    '판매량':[120,90,50,70,80]
}

df=pd.DataFrame(data)
print('전체 메뉴 평균 가격: {}'.format(df['가격'].mean()))

print('판매량이 80개 이상인 메뉴:')
print(df[df['판매량'] >= 80])

print('가장 비싼 메뉴:')
print(df[df['가격'] == df['가격'].max()])

print('가장 저렴한 메뉴:')
print(df[df['가격'] == df['가격'].min()])

print('커피 카테고리 메뉴:')
print(df[df['카테고리'] == '커피'])
```
