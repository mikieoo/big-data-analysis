```python
import pandas as pd

df=pd.Series([781,650,705,406,580,450,550,640],
             index=['A','B','C','D','E','F','G','H'])

df
```

```python
result = df[(df < 500) | (df > 700)]
result
```

```python
result = df[df > df['B']]
result
```

```python
result = df[df < 600].index
print(list(result))
```

```python
# 매출액이 600미만인 팀들의 매출액을 20% 증가
result = df[df < 600] * 1.2
result
```
