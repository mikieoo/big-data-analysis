```python
def apply_operation(numbers, func):
    return [func(n) for n in numbers]

def square(x):
    return x * x

def negate(x):
    return -x

nums = [1, 2, 3, 4]
print("제곱:", apply_operation(nums, square))
print("부호 반전:", apply_operation(nums, negate))
```

```python
class Counter:
    def __init__(self):
        self.count = 0

    def increment(self):
        self.count += 1

    def get_value(self):
        return self.count

c = Counter()
c.increment()
c.increment()
print("카운트:", c.get_value())
```

```python
def check_even(n):
  if n%2==0:
    print('짝수입니다.')
  else:
    print('홀수입니다.')

check_even(4)
```

```python
class Student:
  def __init__(self, name, age):
    self.name=name
    self.age=age

  def introduce(self):
    print('저는 {0}이고, {1}살입니다.'.format(self.name, self.age))

s = Student('지민',18)
s.introduce()
```

```python
def average(a=[]):
  sum = 0
  if len(a) == 0:
    return '데이터 없음'
  for i in a:
    sum += i
  return sum / len(a)

print(average())
```

```python
class Calculator():
  def __init__(self, num1, num2):
    self.num1 = num1
    self.num2 = num2

  def add(self):
    return self.num1 + self.num2

  def divide(self):
    return self.num1 / self.num2

calc = Calculator(10, 2)
print(calc.add())
print(calc.divide())
```

```python
# 사용 예:
# acc = BankAccount("채민", 10000)
# acc.deposit(2000)
# acc.withdraw(5000)
# acc.check_balance() → 7000

class BankAccount():
  def __init__(self, name, money):
    self.money = money
    self.name = name

  def deposit(self, amount):
    self.money += amount

  def withdraw(self, amount):
    self.money -= amount

  def check_balance(self):
    return self.money

acc = BankAccount("채민", 10000)
acc.deposit(2000)
acc.withdraw(5000)
acc.check_balance()
```

```python
def word_count(text):
    text = text.lower()
    text = text.replace(",", "").replace(".", "")
    words = text.split()
    count_dict = {}

    for word in words:
        if word in count_dict:
            count_dict[word] += 1
        else:
            count_dict[word] = 1

    return count_dict
```

```python
from collections import Counter

text = "Hello, hello world."
text = text.lower().replace(",", "").replace(".", "")
words = text.split()

result = Counter(words)
print(result)
# 출력: Counter({'hello': 2, 'world': 1})
```
